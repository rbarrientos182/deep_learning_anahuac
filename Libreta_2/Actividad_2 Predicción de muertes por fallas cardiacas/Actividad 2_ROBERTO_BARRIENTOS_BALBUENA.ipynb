```python
import pandas as pd
import numpy as np
from numpy.random import seed
seed(30) ## No muevas este parámetro, nos aseguramos de que tus resultados (si haces bien el ejercicio) 
         ## estén alineados con las opciones de la plataforma de edX.

from sklearn.preprocessing import MinMaxScaler

import tensorflow
tensorflow.random.set_seed(30) ## No muevas este parámetro, nos aseguramos de que tus resultados (si haces bien el ejercicio) 
                               ## estén alineados con las opciones de la plataforma de edX.

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.optimizers import SGD
    
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv("heart_failure_clinical_records_dataset.csv")
print("El conjunto de datos tiene {} filas y {} columnas".format(df.shape[0],df.shape[1]))
df.head()
```

<ol start=3>
    <li>Genera una muestra probabilística tomando en cuenta todos los datos, para ello tendrás que revolver los elementos. Después, imprime los primeros cinco elementos y responde, <strong>¿Cuál es la edad del primer elemento?</strong></li>
</ol>

```python
df = df.sample(frac=1)
```

```python
print(df.age.head(1))
```

<ol start = 4>
    <li>Separa la variable objetivo de tu conjunto de datos. Recuerda que aquí se definen las variables dependientes e independientes.</li>
</ol>

```python
y = df.DEATH_EVENT
X = df.drop("DEATH_EVENT", axis=1)
X,y
```

<ol start=5>
    <li>Realiza un escalamiento de los datos, imprime la variable "X" &amp; "y" escaladas y contesta lo siguiente: <strong>¿Cuál es el primer elemento del arreglo generado de la variable X?</strong></li>
</ol>

```python
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

X = scaler_X.fit_transform(X)
y = scaler_y.fit_transform(y.values.reshape(-1, 1))
```

```python
print("Primer elemento del arreglo X escalado:")
print(X[0])
```

<ol start=6>
    <li>Realiza la separación de datos en conjuntos de entrenamiento y prueba. Haz una división de tal forma que el 70% de los datos sean de entrenamiento y el 30% de prueba.</li>
</ol>    

```python
split = round(X.shape[0]*0.7)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]
```

<ol start=7>
    <li>Imprime los resultados del split de datos de los conjuntos de entrenamientos y prueba de las variables dependiente e independiente. Responde a lo siguiente: <strong>¿Cuántas filas posee la variable independiente de entrenamiento?</strong></li>
</ol>

```python
print(f"Filas de X_train (independiente de entrenamiento): {X_train.shape[0]}")
print(f"Filas de X_test  (independiente de prueba):       {X_test.shape[0]}")
print(f"Filas de y_train (dependiente de entrenamiento):   {y_train.shape[0]}")
print(f"Filas de y_test  (dependiente de prueba):         {y_test.shape[0]}")
```

```python
X_train.shape[1]
```

<ol start=8>
    <li>Genera un red neuronal utilizando  las librerías de TensorFlow y Keras. La rede debe contar con las siguientes características:</li>
    <ol>
        <li><strong>Cinco capas ocultas.</strong></li>
        <li>Número de neuronas dentro de las capas ocultas: <strong>20</strong></li>
        <li>Función de activación entre las capas ocultas: <strong>Relu</strong></li>
        <li>Función de activación entre la última capa oculta y la capa de salida: <strong>Sigmoide</strong></li>
        <li>Learning rate: <strong>10%</strong></li>
        <li>Optimizador: <strong>Descenso del gradiente estocástico.</strong></li>
        <li>Loss: <strong>Error promedio cuadrado.</strong></li>
        <li>Epochs: <strong>500</strong></li>
        <li>Tamaño del batch: <strong>10</strong></li>
    </ol>
</ol>

```python
model = keras.Sequential([
    layers.Dense(20, activation='relu',input_shape=(12,)),
    layers.Dense(20, activation='relu'),
    layers.Dense(20, activation='relu'),
    layers.Dense(20, activation='relu'),
    layers.Dense(20, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

sgd = SGD(learning_rate=0.1)
model.compile(optimizer=sgd,
              loss='mean_squared_error')

model.fit(X_train, y_train, epochs=500, batch_size=10)
```

<ol start=9>
    <li>Carga la siguiente línea para definir las variables con las etiquetas predichas con el conjunto de entrenamiento y las etiquetas predichas con el conjunto de prueba: </li>
</ol>

```python
y_hat_train = model.predict(X_train).round()
y_hat_test = model.predict(X_test).round()
```

<ol start=10>
    <li>Genera una matriz de confusión que compare las etiquetas reales con las etiquetas predichas.</li>
</ol>

```python
confusion_matrix(y_test,y_hat_test)
```
