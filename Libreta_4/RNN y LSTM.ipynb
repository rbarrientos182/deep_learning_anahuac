# Redes Neuronales Recurrentes (RNN)

<center><img src="img/Rolled_RNN.png" width = "12%"></center>
<p style='text-align: justify;'>Son redes neuronales que cuentan con ciclos (Loops) que les brinda la capacidad de mantener información acerca de eventos previos. Una red neuronal recurrente no es tan diferente a una red neuronal clásica, sino que podemos pensar en ellas como múltiples copias de la misma red, donde cada copia previa le comparte su información a las copias consecuentes.</p>
<center><img src="img/Unrolled_RNN.png" width = "60%"></center>
<p style='text-align: justify;'>La naturaleza de cadena demuestra que las RNN están estrechamente relacionadas a listas y secuencias, son la estructura natural de red neuronal para manipular este tipo de datos. De hecho, en los últimos años se han desarrollado diversas aplicaciones de este tipo de redes como modelaje de lengua, traducción, captura de imágenes y videos.</p>
<p style='text-align: justify;'>Uno de los puntos mas atractivos de las redes neuronales recurrentes es que pueden conectar la información pasada a la tarea presente. Algunas veces, solo es necesario mirar a la información más reciente para poder realizar la tarea presente, pero hay ocasiones en donde se necesita un mayor contexto. Desafortunadamente, a medida que crece la brecha entre la adquisición de la información relevante y su uso, las RNN pierden su habilidad de aprender a conectar la información.</p>

# Redes de Memoria Larga a Corto Plazo (LSTM)

<p style='text-align: justify;'>Son un tipo especial de RNN capaces de aprender dependencias a largo plazo y que cuentan con la habilidad de recordar información por largos periodos de tiempo.</p>
<center><img src="img/LSTM_layers.png" width = "80%"></center>
<p style='text-align: justify;'>La clave principal de la memoria larga-corto plazo es el estado de la celda, la cual es representada por la línea horizontal que recorre la parte superior del diagrama. Además, las redes LSTM tienen una unidad de memoria que contiene tres puertas que controlan el modo en que la información fluye dentro o fuera de la unidad.</p>
<center><img src="img/LTSM_Explain1.png" width = "50%"></center>

1) La puerta del olvido brinda a la celda la capacidad de olvidar información almacenada y le permite discernir entre datos importantes y datos superfluos. La decisión es llevada a cabo por medio de una función sigmoide que devuelve un valor entre 0 y 1 para cada estado de la celda previos, el valor de cero significa que se deshará del valor completamente mientras que el valor de uno significa que se quedará con la información en su totalidad.
<center><img src="img/LTSM_Explain2.png" width = "50%"></center>

2) La puerta de entrada regula la entrada de nueva información hacia la memoria de la celda utilizando una función sigmoide en conjunto con la función tanh que crea un vector de nueva información candidata 𝐶 ̃que puede o no ser agregado a la celda.
<center><img src="img/LTSM_Explain3.png" width = "50%"></center>

3) La puerta de salida filtra la información que devolverá la celda de procesamiento. El proceso inicia utilizando una función sigmoide que decide cuáles segmentos de la celda formarán parte de la salida. Posteriormente, se aplica la función tanh al estado actual de la celda la cual se multiplica a la salida de la función sigmoide, de manera que solamente salen los segmentos que fueron escogidos.
